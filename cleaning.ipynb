```python
import pandas as pd

df = pd.read_csv("data/day2.csv")
print("Original Data:")
print(df)

print("\n Missing Values:")
print(df.isnull().sum())


df["Marks"] = df["Marks"].fillna(0)
df["Age"] = df["Age"].fillna(df["Age"].median())


print("\n remove duplicates")
clean = df.drop_duplicates()
print(clean)

```
